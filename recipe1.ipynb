```python
import pandas as pd 
import matplotlib.pyplot as plt
df=pd.read_csv(r"C:\Users\Dell\Downloads\recipes (1).csv")
'print(df)
```

```python
top_10 = df.nlargest(10, "rating")

plt.figure(figsize=(9, 8))
plt.bar(top_10["recipe_name"], top_10["rating"])

plt.xticks(rotation=50, ha="right")
plt.title("Top 10 Highest Rated Recipes")
plt.xlabel("Recipe Name")
plt.ylabel("Rating")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df["rating"].dropna(),bins=10,edgecolor="black")
plt.title("Distribution of recipe rating")
plt.xlabel("Rating")
plt.ylabel("number of recipes")
plt.show()
```

```python
top_5=df["cuisine_path"].value_counts().head(5)
plt.figure(figsize=(7,7))
plt.pie(
    top_5,
    labels=top_5.index,
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Top 5 Cuisine by recipe count")
plt.show()
```

```python
top_10_cuisines=df["cuisine_path"].value_counts().head(10)
plt.figure(figsize=(9,5))
plt.barh(top_10_cuisines.index,top_10_cuisines.values)
plt.title("Top 10 Cuisines")
plt.xlabel("Number of recipes")
plt.ylabel("Cuisine")
plt.show()
```

```python
plt.figure(figsize=(7,5))
plt.boxplot(df["rating"].dropna())
plt.title("Distribution of recipe rating")
plt.ylabel("Rating")
plt.show()
```

```python
top_5_cuisines=df["cuisine_path"].value_counts().head(5).index
avg_rating=(
    df[df["cuisine_path"].isin(top_5_cuisines)]
    .groupby("cuisine_path")["rating"]
    .mean()
)
plt.figure(figsize=(9,5))
plt.bar(avg_rating.index,avg_rating.values)
plt.title("Average rating of top 5 cuisines")
plt.ylabel("average rating")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
